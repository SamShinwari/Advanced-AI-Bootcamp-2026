#Element-wise operations

```python
import tensorflow as tf
```

```python

a = tf.constant([10, 20, 30])
b = tf.constant([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

#tf.reduce_sum() — Calculate the sum

```python
x = tf.constant([10, 20, 30, 40])

result = tf.reduce_sum(x)

print(result)
```

#tf.reduce_mean() — Calculate the average

```python
x = tf.constant([10.0, 20.0, 30.0, 40.0])

result = tf.reduce_mean(x)

print(result)
```

#tf.argmax() — Find the index of the largest value

```python
predictions = tf.constant([0.1, 0.7, 0.2])

result = tf.argmax(predictions)

print(result)
```

#tf.matmul() — Matrix multiplication

```python
A = tf.constant([
    [1, 2],
    [3, 4]
])

B = tf.constant([
    [5, 6],
    [7, 8]
])

result = tf.matmul(A, B)

print(result)
```
